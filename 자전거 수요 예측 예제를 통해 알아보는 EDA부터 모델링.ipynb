## 15. 선형회귀 모델 학습

시간, 날씨, 날짜 정보를 바탕으로 대여량을 예측하는 선형회귀 모델을 학습합니다.

### 선형회귀란?

선형회귀는 입력값에 계수를 곱하고 절편을 더해 숫자를 예측하는 모델입니다. 가장 단순한 형태는 `y = ax + b`입니다. 여기서 `x`는 입력값, `y`는 예측값, `a`는 계수, `b`는 절편입니다.

예를 들어 `예상 대여량 = 12 * hour + 30`이라면 `hour`의 계수는 12입니다. `hour`가 1 증가할 때 모델의 예측값이 평균적으로 12만큼 증가한다는 뜻입니다.

입력 변수가 여러 개라면 `y = a1x1 + a2x2 + ... + b`처럼 각 변수마다 계수가 하나씩 붙습니다. 시간, 기온, 습도 같은 변수에 각각 계수를 붙이고, 그 합이 실제 대여량과 최대한 비슷해지도록 계수를 찾습니다.

모델 구조가 단순해서 결과를 해석하기 쉽습니다. 어떤 변수의 계수가 양수이면 예측값을 올리는 방향으로, 음수이면 예측값을 낮추는 방향으로 작용합니다.

## 17. 선형회귀 계수 확인

선형회귀는 각 변수에 계수를 하나씩 부여합니다. 계수의 절댓값이 크면 예측값을 움직이는 영향이 큰 변수로 볼 수 있습니다. 다만 숫자형 변수와 더미 변수가 함께 들어 있으므로 계수의 크기만으로 원인을 단정하지는 않습니다.

````python
# 선형회귀 계수를 표로 정리하세요.
coef_df = pd.DataFrame({
    "특징": X_train.columns,
    "계수": model.coef_
})
coef_df["계수_절댓값"] = coef_df["계수"].abs()
coef_df = coef_df.sort_values("계수_절댓값", ascending=False)
coef_df.head(10)
````

````python
# 절댓값 기준 상위 계수를 그래프로 확인하세요.
top_coef = coef_df.head(15).sort_values("계수")

sns.barplot(data=top_coef, x="계수", y="특징", color="#4C78A8")
plt.axvline(0, color="gray", linewidth=1)
plt.title("선형회귀 계수")
plt.xlabel("계수")
plt.ylabel("특징")
plt.tight_layout()
plt.show()
````
