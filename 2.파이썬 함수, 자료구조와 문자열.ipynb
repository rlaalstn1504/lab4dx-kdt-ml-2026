## 1. 함수
### 1) 함수란 무엇인가?
- **반복되는 코드를 묶어서 이름을 붙여둔 것**
- 한 번 정의해두면 필요할 때마다 **호출(call)** 해서 재사용 가능
- 코드를 **짧고 간결**하게 만들고, **유지보수**도 쉬워짐

### 2) 함수 정의와 호출

#### 기본 문법
```python
def 함수이름(매개변수):
    실행할 코드
    return 결과값
```

- `def` : 함수를 정의할 때 사용하는 키워드

- `매개변수(parameter)` : 함수에 전달되는 값(선택적)

- `return`: 결과를 함수 바깥으로 돌려줌 (없으면 None 반환)

#### 자주 하는 실수

return을 빼먹고 함수 내부에서 출력만 함  
→ 출력은 되지만 결과값은 None 반환

함수 정의만 하고 호출을 안 함  
→ 함수는 실행되지 않음

들여쓰기 오류  
→ 함수 내부 코드는 반드시 들여쓰기 필요


````python
# 예시
def add(x, y):
    return x + y

print(add(3, 5))  # 문자열을 정수로 바꿉니다.
print(add(10, 20))  # 문자열을 정수로 바꿉니다.

def greet(name):
    return f"안녕하세요, {name}님!"

print(greet("철수"))  # 문자열을 정수로 바꿉니다.
````

### 3) 익명 함수 (lambda)

이름 없는 간단한 함수 정의할 때 사용

lambda 매개변수: 표현식

````python
# 예시
square = lambda x: x**2
print(square(5))   # 25
````

### 4) 여러 인자를 받는 *, ** 문법

파이썬에서는 함수가 가변 개수의 인자를 받을 수 있도록 *args, **kwargs 문법을 제공합니다.

`*args`

여러 개의 위치 인자를 튜플(tuple) 형태로 받음

인자의 개수를 미리 정하지 않아도 됨

```python
def add_all(*args):
    return sum(args)

print(add_all(1, 2, 3))       # 6
print(add_all(4, 5, 6, 7, 8)) # 30
```

`**kwargs`

여러 개의 키워드 인자를 딕셔너리(dict) 형태로 받음

함수 호출 시 key=value 형식으로 전달
```python
def print_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} : {value}")  # 문자열을 정수로 바꿉니다.

print_info(name="Alice", age=25, city="Seoul")
# name : Alice
# age : 25
# city : Seoul
```

👉 *args와 **kwargs를 함께 쓰면, 위치 인자와 키워드 인자를 동시에 유연하게 처리할 수 있음.

## 2. 리스트, 딕셔너리, 튜플

### 1) 리스트(List)
- 여러 값을 하나의 변수에 순서대로 저장할 수 있는 자료형
- 대괄호 `[ ]`로 묶어 표현
- 순서(인덱스)가 있어서 **위치로 접근** 가능 (인덱스는 0부터 시작)

#### 주요 메서드
- `append(값)` : 맨 뒤에 값 추가
- `insert(위치, 값)` : 원하는 위치에 값 삽입
- `remove(값)` : 해당 값을 찾아 삭제 (첫 번째 것만 삭제)
- `pop([위치])` : 특정 위치의 값을 꺼내고 삭제 (인덱스 미지정 시 마지막 원소 제거)
- `count(값)` : 리스트 안에서 해당 값의 개수를 반환
- `sort()` : 오름차순 정렬 (내림차순 정렬은 sort(reverse=True))

#### 인덱싱(Indexing)과 슬라이싱(Slicing)

인덱싱 : 특정 위치(인덱스)의 값을 가져오기 → 리스트[위치]

슬라이싱 : 리스트의 일부 구간을 잘라내기 → 리스트[시작:끝:간격]


````python
# 리스트 생성
fruits = ["apple", "banana", "cherry", "apple", "banana", "grape"]
print("처음 리스트:", fruits)  # 문자열을 정수로 바꿉니다.
````

````python
# 인덱스로 접근 (Indexing)
print("첫 번째 원소:", fruits[0])    # apple
print("마지막 원소:", fruits[-1])   # grape
````

````python
# 슬라이싱 (Slicing)
print("0~2번 인덱스:", fruits[0:3])   # ['apple', 'banana', 'cherry']
print("2번 이후:", fruits[2:])       # ['cherry', 'apple', 'banana', 'grape']
print("0~끝까지 2칸 간격:", fruits[::2])  # ['apple', 'cherry', 'banana']
````

````python
# append : 맨 뒤에 추가
fruits.append("melon")
print("append 후:", fruits)  # 문자열을 정수로 바꿉니다.
````

````python
# insert : 원하는 위치에 삽입
fruits.insert(1, "orange")
print("insert 후:", fruits)  # 문자열을 정수로 바꿉니다.
````

````python
# remove : 특정 값 삭제
fruits.remove("banana")  # 첫 번째 "banana" 삭제
print("remove 후:", fruits)  # 문자열을 정수로 바꿉니다.
````

````python
# pop : 외부로 값 꺼내기
item = fruits.pop()   # 마지막 원소 제거
print("pop으로 꺼낸 값:", item)  # 문자열을 정수로 바꿉니다.
print("pop 후:", fruits)  # 문자열을 정수로 바꿉니다.
````

````python
# del : 특정 위치 삭제
del fruits[0]    # 0번 인덱스 삭제
print("del 후:", fruits)  # 문자열을 정수로 바꿉니다.
````

````python
# count : 특정 값 개수 세기
print("apple 개수:", fruits.count("apple"))  # 문자열을 정수로 바꿉니다.
````

````python
# sort : 정렬
numbers = [5, 2, 9, 1, 7]
numbers.sort()  # 리스트를 지정한 기준으로 정렬합니다.
print("오름차순 정렬:", numbers)  # 문자열을 정수로 바꿉니다.
numbers.sort(reverse=True)  # 리스트를 지정한 기준으로 정렬합니다.
print("내림차순 정렬:", numbers)  # 문자열을 정수로 바꿉니다.
````

### 2) 딕셔너리(Dictionary)
- 데이터를 **키(key) : 값(value)** 형태로 저장
- 중괄호 `{ }`로 표현
- 키를 이용해 값에 접근

#### 특징
- 리스트는 "순서"로 값에 접근, 딕셔너리는 "이름(키)"로 값에 접근
- 키는 고유해야 하며, 중복 불가


````python
# 딕셔너리 생성
student = {"name": "홍길동", "age": 20, "major": "컴퓨터공학"}
print("딕셔너리 전체:", student)  # 문자열을 정수로 바꿉니다.

# 키로 값 접근
print("이름:", student["name"])  # 문자열을 정수로 바꿉니다.

# 값 수정
student["age"] = 21
print("나이 수정 후:", student)  # 문자열을 정수로 바꿉니다.

# 새로운 키-값 추가
student["grade"] = "A"
print("새 항목 추가 후:", student)  # 문자열을 정수로 바꿉니다.

# 특정 키 삭제
del student["major"]
print("삭제 후:", student)  # 문자열을 정수로 바꿉니다.
````

### 3) 튜플(Tuple)
- 리스트와 비슷하지만, **한 번 정의하면 변경 불가능(immutable)**  
- 소괄호 `( )`로 묶어 표현  
- 값 수정, 추가, 삭제 불가 → 주로 고정된 데이터를 저장할 때 사용  

```python
# 튜플 생성
numbers = (1, 2, 3, 4)

# 인덱스로 접근
print("첫 번째 원소:", numbers[0])   # 1

# 변경 불가능 확인
# numbers[0] = 10   # 오류 발생 (튜플은 수정 불가)

# 튜플 언패킹
a, b, c, d = numbers
print("언패킹:", a, b, c, d)  # 문자열을 정수로 바꿉니다.
```

### ✅ 핵심 정리
- **리스트** : `[ ]`, 순서가 있고 인덱스로 접근  
  → 주요 메서드 : `append`, `insert`, `remove`  

- **딕셔너리** : `{ }`, 키-값 쌍으로 저장, 키로 접근  
  → 수정, 추가, 삭제 가능  

- **튜플** : `( )`, 순서가 있으나 **값 변경 불가**  
  → 주로 변경되면 안 되는 데이터 저장

### 1) 문자열이란?
- 문자들을 나열한 데이터, `" "` 또는 `' '` 로 묶어 표현
- 예: `"Hello"`, `'Python'`
- 문자열도 **인덱스(위치 번호)** 로 접근 가능 (0부터 시작)


### 2) 인덱싱(Indexing)과 슬라이싱(Slicing)
- **인덱싱**: 문자열에서 특정 위치의 문자 1개를 꺼냄
- **슬라이싱**: 문자열의 일부분을 잘라서 가져옴

```python
text = "Python"
print(text[0])     # 'P' (0번 인덱스)
print(text[-1])    # 'n' (뒤에서 첫 글자)
print(text[0:4])   # 'Pyth' (0번부터 3번까지)
print(text[2:])    # 'thon' (2번부터 끝까지)
print(text[:3])    # 'Pyt' (처음부터 2번까지)
```

| `join(list)`    | 리스트를 특정 구분자로 이어 문자열로 반환    | `"-".join(["a","b","c"])` → `"a-b-c"`     |

### 퀴즈) 나만의 단어장 저장하기

사용자에게 영어 단어와 뜻을 입력받아 `vocabulary.txt`에 한 줄씩 저장하는 프로그램을 만들어 봅시다.

조건:
- 파일은 추가 모드(`a`)로 엽니다.
- 저장 형식은 `영어단어: 뜻`입니다.
- 영어 단어 입력에서 `q`가 들어오면 종료합니다.
- 뜻 입력에서 `q`가 들어와도 종료합니다.


````python
with open("vocabulary.txt", "a", encoding="utf-8") as f:
    while True:
        eng = input("영어 단어: ").strip()  # 양쪽 공백을 제거합니다.
        if eng == "q":
            break

        meaning = input("뜻: ").strip()  # 양쪽 공백을 제거합니다.
        if meaning == "q":
            break

        f.write(f"{eng}: {meaning}\n")
````

### 퀴즈) 단어장으로 미니 퀴즈 만들기

`vocabulary.txt`에 `영어단어: 뜻` 형식으로 저장된 내용을 한 줄씩 읽어, 뜻을 보여주고 영어 단어를 맞히는 프로그램을 만들어 봅시다.

조건:
- 파일의 각 줄을 읽어 영어 단어와 뜻으로 나눕니다.
- 사용자에게 뜻을 보여주고 영어 단어를 입력받습니다.
- 정답이면 `맞았습니다!`, 오답이면 정답을 함께 출력합니다.


````python
with open("vocabulary.txt", "r", encoding="utf-8") as f:
    for line in f:
        eng, meaning = line.strip().split(": ", 1)  # 문자열을 기준에 따라 나눕니다.
        answer = input(f"{meaning}: ").strip()  # 양쪽 공백을 제거합니다.

        if answer.lower() == eng.lower():
            print("맞았습니다!\n")  # 문자열을 정수로 바꿉니다.
        else:
            print(f"아쉽습니다. 정답은 {eng}입니다.\n")  # 문자열을 정수로 바꿉니다.
````

### 퀴즈) 랜덤 단어 퀴즈 만들기

`vocabulary.txt` 내용을 딕셔너리로 읽어 둔 뒤, 단어를 무작위로 하나씩 내는 퀴즈 프로그램을 만들어 봅시다.

조건:
- 파일 내용을 딕셔너리에 `영어단어: 뜻` 형태로 저장합니다.
- 문제는 랜덤으로 출제합니다.
- 사용자가 `q`를 입력하면 종료합니다.
- 정답/오답 메시지를 출력합니다.


````python
import random

vocab = {}
with open("vocabulary.txt", "r", encoding="utf-8") as f:
    for line in f:
        eng, meaning = line.strip().split(": ", 1)  # 문자열을 기준에 따라 나눕니다.
        vocab[eng] = meaning

eng_words = list(vocab.keys())

while True:
    eng = random.choice(eng_words)
    meaning = vocab[eng]
    answer = input(f"{meaning}: ").strip()  # 양쪽 공백을 제거합니다.

    if answer == "q":
        break

    if answer.lower() == eng.lower():
        print("맞았습니다!\n")  # 문자열을 정수로 바꿉니다.
    else:
        print(f"아쉽습니다. 정답은 {eng}입니다.\n")  # 문자열을 정수로 바꿉니다.
````

### 문제 5. 단어 빈도 세기
사용자로부터 문장을 입력받아, 단어별 등장 횟수를 세어 출력하세요.  
예: `"I love Python and I love coding"` →  
`I: 2, love: 2, Python: 1, and: 1, coding: 1`

<details><summary>정답</summary>

```python
# 풀이 흐름: 문자열을 다루기 쉬운 형태로 정리합니다.
# 필요한 값만 추출하거나 비교합니다.

sentence = input("문장을 입력하세요: ")  # 분석할 문장을 입력받습니다.

words = sentence.split()  # 공백 기준으로 단어 분리
freq = {}  # 빈 딕셔너리 생성

for w in words:  # 값을 하나씩 꺼내 반복합니다.
    w = w.lower()  # 대소문자 구분하지 않도록 소문자로 변환
    if w in freq:  # 조건을 확인합니다.

    else:  # 위 조건에 모두 해당하지 않을 때 실행합니다.
        freq[w] = 1  # 단어별 등장 횟수를 누적합니다.

# 결과 출력
for word, count in freq.items():  # 값을 하나씩 꺼내 반복합니다.
    print(f"{word}: {count}")  # 결과를 화면에 출력합니다.
```

### 선택(심화) 1. 자주 나온 단어 상위 3개 찾기

문자열을 받아 단어 빈도를 세고, 가장 많이 나온 단어 3개를 반환하는 함수를 만들어 보세요.
대소문자는 무시하고 쉼표와 마침표는 제거해도 됩니다.

<details>
<summary>정답 보기</summary>

```python
# 풀이 흐름: 문자열을 다루기 쉬운 형태로 정리합니다.
# 필요한 값만 추출하거나 비교합니다.

def top_words(text, n=3):  # 함수를 정의합니다.
    cleaned = text.lower().replace(",", "").replace(".", "")  # 비교하기 쉽도록 문자열을 정리합니다.
    counts = {}  # 값별 개수를 저장할 딕셔너리입니다.

    for word in cleaned.split():  # 값을 하나씩 꺼내 반복합니다.
        counts[word] = counts.get(word, 0) + 1  # 단어별 등장 횟수를 누적합니다.

    ranked = sorted(counts.items(), key=lambda x: (-x[1], x[0]))  # 정렬된 새 리스트를 만듭니다.
    return ranked[:n]  # 계산한 값을 함수 밖으로 돌려줍니다.


sample = "Python is fun. Python is easy, and python is powerful."  # 파일에 저장할 예시 문자열입니다.
print(top_words(sample))  # 결과를 화면에 출력합니다.

```

```python
# 풀이 흐름: 문자열을 다루기 쉬운 형태로 정리합니다.
# 필요한 값만 추출하거나 비교합니다.

def is_sentence_palindrome(text):  # 함수를 정의합니다.
    cleaned = "".join(ch.lower() for ch in text if ch.isalnum())  # 비교하기 쉽도록 문자열을 정리합니다.
    return cleaned == cleaned[::-1]  # 계산한 값을 함수 밖으로 돌려줍니다.


print(is_sentence_palindrome("Never odd or even"))  # 결과를 화면에 출력합니다.
```

### 선택(심화) 3. 점수 파일을 읽어 평균 구하기

아래 형식의 파일을 읽어 이름과 점수를 딕셔너리로 만들고, 평균 점수까지 구해 보세요.

```text
민수,90
지민,85
서연,95
```

<details>
<summary>정답 보기</summary>

```python
# 풀이 흐름: 예시 파일을 만들고 다시 읽어옵니다.
# 읽어온 값을 딕셔너리로 정리한 뒤 평균을 계산합니다.

sample = "민수,90\n지민,85\n서연,95\n"  # 파일에 저장할 예시 문자열입니다.

with open("scores.txt", "w", encoding="utf-8") as f:  # 파일을 열고 사용이 끝나면 자동으로 닫습니다.
    f.write(sample)


def load_scores(filename):  # 함수를 정의합니다.
    scores = {}  # 이름과 점수를 저장할 딕셔너리입니다.

    with open(filename, "r", encoding="utf-8") as f:  # 파일을 열고 사용이 끝나면 자동으로 닫습니다.
        for line in f:  # 값을 하나씩 꺼내 반복합니다.
            name, score = line.strip().split(",")  # 문자열을 기준에 따라 나눕니다.
            scores[name] = int(score)  # 문자열을 정수로 바꿉니다.

    return scores  # 계산한 값을 함수 밖으로 돌려줍니다.


scores = load_scores("scores.txt")  # 이름과 점수를 저장할 딕셔너리입니다.
average = sum(scores.values()) / len(scores)  # 평균 점수를 계산합니다.

print(scores)  # 결과를 화면에 출력합니다.
print("평균:", average)  # 결과를 화면에 출력합니다.
```
